```python
from flask import Flask, render_template, request, send_file
import json
import re

app = Flask(__name__)

# بارگذاری دیتابیس جدید
with open("diseases.json", "r", encoding="utf-8") as f:
    data = json.load(f)
    # دریافت لیست از کلید ear_codes
    diseases = data.get("ear_codes", [])

def clean_and_split(text):
    if not text:
        return set()
    text = text.lower()
    text = text.replace("ي", "ی").replace("ك", "ک")
    words = re.findall(r'[a-zA-Z0-9آ-ی‌]+', text)
    return set(words)

def guess_code(user_text):
    user_words = clean_and_split(user_text)
    if not user_words:
        return None

    best_match = None
    max_matches = 0

    for item in diseases:
        # ترکیب کردن کلمات کلیدی، مترادف‌ها و عنوان‌ها
        title_words = clean_and_split(item.get('title', ''))
        keywords_words = clean_and_split(" ".join(item.get('keywords', [])))
        synonyms_words = clean_and_split(" ".join(item.get('synonyms', [])))

        # مجموع تمام واژه‌های مربوط به این آیتم دیتابیس
        all_item_words = title_words.union(keywords_words).union(synonyms_words)

        # کلمات مشترک
        common_words = user_words.intersection(all_item_words)

        # حذف حروف پر تکرار و بی‌ارزش
        stop_words = {'و', 'در', 'به', 'از', 'که', 'با', 'بیمار', 'است', 'شدید', 'is', 'the', 'and', 'with', 'of', 'for', 'to'}
        valid_common = common_words - stop_words

        if len(valid_common) > max_matches:
            max_matches = len(valid_common)
            best_match = item

    return best_match if max_matches > 0 else None

@app.route('/')
def home():
    return render_template('index.html')

@app.route('/about')
def about():
    return render_template('about.html')

@app.route('/search', methods=['GET', 'POST'])
def search():
    result = None
    if request.method == 'POST':
        symptoms = request.form.get('symptoms', '')
        result = guess_code(symptoms)
        if result is None:
            result = False
    return render_template('search.html', result=result)

@app.route('/download')
def download():
    return send_file('diseases.json', as_attachment=True)

# اجرای وب‌سایت
import threading
threading.Thread(target=app.run, kwargs={'port': 5000}).start()
```
